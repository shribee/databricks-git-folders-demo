```sql
create or replace temporary view order_items as
values
(1, array('smartphone', 'laptop', 'monitor')),
(2, array('tablet', 'headphones', 'smartwatch')),
(3, array('keyboard', 'mouse'))
as orders(order_id, items);
```

```sql
select *
from order_items
```

```sql
select explode(items) as item
from order_items
```

## 1. TRANSFORM higher order function

```sql
select 
transform(items, item -> upper(item)) as upper_items
from order_items
```

## 2. FILTER higher order function

```sql
select 
filter(items, item -> item like '%smart%') as smart_items
from order_items
```

## 3. EXISTS higher order function

```sql
select 
exists(items, item -> item = 'monitor') as smart_items
from order_items
```

```sql
create or replace temporary view order_items_new as
values
(1, array(
          named_struct('name', 
            'smartphone', 
            'price',
            699
          )
          ,
          named_struct(
            'name', 
            'laptop', 
            'price',
            1199
           )
           ,
          named_struct(
            'name',
            'monitor',
            'price',
            399
           )
))
,
(2, array(
          named_struct('name', 
            'tablet', 
            'price',
            599
          ),
          named_struct(
            'name', 
            'headphones', 
            'price',
            199
          ),
          named_struct(
            'name',
            'smartwatch',
            'price',
            299
          )
)),
(3, array(
          named_struct(
            'name', 
            'keyboard', 
            'price',
            89
          ),
          named_struct(
            'name', 
            'mouse', 
            'price',
            59
          )
))
as orders(order_id, items);
```

```sql
SELECT * FROM order_items_new
```

```sql
select transform(items, x -> named_struct('name', upper(x.name), 'price', round(x.price*1.1, 2))) AS order_items_new_transformed 
from order_items_new
```

## 4. AGGREGATE higher order function

```sql
select aggregate(items, 0, (acc, x) -> acc + x.price) as total_order_price
from order_items_new
```
