```sql
create or replace function gizmobox.default.get_fullname(fname string, lname string)
returns string
return concat(initcap(fname), ' ', initcap(lname));
```

```sql
select gizmobox.default.get_fullname('john', 'smith');
```

```sql
desc function extended gizmobox.default.get_fullname
```

```sql
select 
payment_id,
order_id,
payment_timestamp,
cast(date_format(payment_timestamp, 'yyyy-MM-dd') as date) as payment_date,
date_format(payment_timestamp, 'HH:mm:ss') as payment_time,
payment_status,
case payment_status
when 1 then 'Succeeded'
when 2 then 'Pending'
when 3 then 'Cancelled'
when 4 then 'Failed'
end as payment_status_values,
payment_method
from gizmobox.bronze.payments
```

```sql
create or replace function gizmobox.bronze.get_payment_status_str(payment_status int)
returns string
return 
case payment_status
when 1 then 'Succeeded'
when 2 then 'Pending'
when 3 then 'Cancelled'
when 4 then 'Failed'
end
```

```sql
select gizmobox.bronze.get_payment_status_str(1);
```

```sql
select 
payment_id,
order_id,
payment_timestamp,
cast(date_format(payment_timestamp, 'yyyy-MM-dd') as date) as payment_date,
date_format(payment_timestamp, 'HH:mm:ss') as payment_time,
payment_status,
gizmobox.bronze.get_payment_status_str(payment_status) as payment_status_values,
payment_method
from gizmobox.bronze.payments
```
