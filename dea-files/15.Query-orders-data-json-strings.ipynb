## 1. to get to element inside json type string, use :

```sql
select 
value,
value:order_id
from gizmobox.bronze.v_orders
```

```sql
select 
value,
value:order_id as order_id,
value:items as items
from gizmobox.bronze.v_orders
```

```sql
select 
value,
value:order_id as order_id,
value:items as items,
value:items[0] as item_1,
value:items[1] as item_2
from gizmobox.bronze.v_orders
```

## 2. to get to element inside struct, dict etc., use . or :

```sql
select 
value,
value:order_id as order_id,
value:items as items,
value:items[0] as item_1,
value:items[1] as item_2,
value:items[0].item_id as item_1_item_id,
value:items[0]:item_id as item_1_item_id2
from gizmobox.bronze.v_orders
```

## 3. to cast field to a diff data type, use ::

```sql
select 
value,
value:order_id as order_id,
value:items as items,
value:items[0] as item_1,
value:items[1] as item_2,
value:items[0].item_id::integer as item_1_item_id
from gizmobox.bronze.v_orders
```

```sql
select 
  value:order_id::Integer as order_id
  , value:items as items
  , value:items[0] as item_1
  , value:items[0].item_id::Integer as item_1_casted
  , value:items[0]:details as item_1_details
  , value:items[0].details.brand as item_1_details_brand
from gizmobox.bronze.v_orders
```

### Can use . notation or :, but for casting ::

```sql
select 
  value:order_id::Integer as order_id
  , value:items as items
  , value:items[0] as item_1
  , value:items[0].item_id::Integer as item_1_casted
  , value:items[0]:details as item_1_details
  , value:items[0]:details:brand as item_1_details_brand
from gizmobox.bronze.v_orders
```
