## 1. extract substring using split function

https://learn.microsoft.com/en-us/azure/databricks/sql/language-manual/functions/split

```sql
select 
    refund_id,
    payment_id,
    refund_timestamp,
    refund_amount,
    refund_reason,
    split(refund_reason,':')[0] as refund_reason_actual,
    split(refund_reason,':')[1] as refund_initiator
from hive_metastore.bronze.refunds
```

## 2. extract substring using egexp_extract function

https://learn.microsoft.com/en-us/azure/databricks/sql/language-manual/functions/regexp_extract

```sql
select 
    refund_id,
    payment_id,
    refund_timestamp,
    refund_amount,
    refund_reason,
    split(refund_reason,':')[0] as refund_reason_actual,
    split(refund_reason,':')[1] as refund_refund_initiator,
    regexp_extract(refund_reason, '^([^:]+):',1) as refund_reason_actual_regex,
    regexp_extract(refund_reason, '^[^:]+:(.*)$', 1) as refund_initiator_regex
from hive_metastore.bronze.refunds
```

## 3. extract date and time

```sql
select 
    refund_id,
    payment_id,
    refund_timestamp,
    cast(date_format(refund_timestamp, "yyyy-MM-dd") as date) as refund_date,
    date_format(refund_timestamp, "HH:mm:ss") as refund_time,
    refund_amount,
    refund_reason,
    split(refund_reason,':')[0] as refund_reason_actual,
    split(refund_reason,':')[1] as refund_refund_initiator,
    regexp_extract(refund_reason, '^([^:]+):',1) as refund_reason_actual_regex,
    regexp_extract(refund_reason, '^[^:]+:(.*)$', 1) as refund_initiator_regex
from hive_metastore.bronze.refunds
```

## 4. write result to silver schema in hive_metastore

```sql
use catalog hive_metastore;
create schema if not exists hive_metastore.silver;
```

```sql
create or replace table hive_metastore.silver.refunds as
select 
    refund_id,
    payment_id,
    refund_timestamp,
    cast(date_format(refund_timestamp, "yyyy-MM-dd") as date) as refund_date,
    date_format(refund_timestamp, "HH:mm:ss") as refund_time,
    refund_amount,
    refund_reason,
    split(refund_reason,':')[0] as refund_reason_actual,
    split(refund_reason,':')[1] as refund_refund_initiator,
    regexp_extract(refund_reason, '^([^:]+):',1) as refund_reason_actual_regex,
    regexp_extract(refund_reason, '^[^:]+:(.*)$', 1) as refund_initiator_regex
from hive_metastore.bronze.refunds
```
