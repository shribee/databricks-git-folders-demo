## 1. deduplicate array elements

https://learn.microsoft.com/en-us/azure/databricks/sql/language-manual/functions/array_distinct

```sql
select 
json_value.customer_id,
json_value.order_id,
json_value.order_date,
json_value.order_status,
json_value.payment_method,
json_value.total_amount,
json_value.transaction_timestamp,
array_distinct(json_value.items) as items_deduped
from gizmobox.silver.orders_json
```

## 2. explode arrays

https://learn.microsoft.com/en-us/azure/databricks/sql/language-manual/functions/explode

```sql
select 
json_value.customer_id,
json_value.order_id,
json_value.order_date,
json_value.order_status,
json_value.payment_method,
json_value.total_amount,
json_value.transaction_timestamp,
explode(array_distinct(json_value.items)) as item
from gizmobox.silver.orders_json
```

## 3. drill-down into individual item

```sql
create or replace temporary view tv_orders_exploded
as
select 
json_value.customer_id,
json_value.order_id,
json_value.order_date,
json_value.order_status,
json_value.payment_method,
json_value.total_amount,
json_value.transaction_timestamp,
explode(array_distinct(json_value.items)) as item
from gizmobox.silver.orders_json
```

```sql
select 
  tv.customer_id,
tv.order_id,
tv.order_date,
tv.order_status,
tv.payment_method,
tv.total_amount,
tv.transaction_timestamp,
  tv.item.category,
  tv.item.item_id,
  tv.item.name,
  tv.item.price,
  tv.item.quantity,
  tv.item.details.brand,
  tv.item.details.color

from tv_orders_exploded tv
```

```sql
create table if not exists gizmobox.silver.orders_final 
as
select 
  tv.customer_id,
tv.order_id,
tv.order_date,
tv.order_status,
tv.payment_method,
tv.total_amount,
tv.transaction_timestamp,
  tv.item.category,
  tv.item.item_id,
  tv.item.name,
  tv.item.price,
  tv.item.quantity,
  tv.item.details.brand,
  tv.item.details.color

from tv_orders_exploded tv
```
